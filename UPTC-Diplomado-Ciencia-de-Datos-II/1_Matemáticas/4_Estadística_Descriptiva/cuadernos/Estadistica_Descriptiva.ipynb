**<h2 id="section1" style="color: #007ACC;">1. Introducción a la Estadística Descriptiva</h2>**

---
Estadística Descriptiva y Definiciones

---

La **estadística** es una disciplina que se encarga de recolectar, organizar, analizar, interpretar y presentar datos. Es una herramienta fundamental en la investigación científica, ya que permite extraer conclusiones a partir de datos observacionales o experimentales. La estadística se utiliza en diversos campos como la ciencia, la economía, la ingeniería, la medicina, y las ciencias sociales, entre otros. Dentro de la estadística como campo de las matemáticas, encontramos dos tipos de estadística que poseen perspectivas y objetivos distintos: la **estadística descriptiva** y la **estadística inferencial**.

La **estadística descriptiva** se enfoca en *resumir y describir las características fundamentales de un conjunto de datos*. Esto incluye técnicas para organizar y visualizar los datos, así como medidas numéricas llamadas **estadísticos** que describen aspectos específicos de los datos, como la tendencia central, la dispersión y la forma de la distribución. Los objetivos principales de la estadística descriptiva son:

- **Resumir**: Proveer resúmenes numéricos y gráficos que faciliten la comprensión de los datos.
- **Describir**: Caracterizar la distribución de los datos mediante medidas o estadísticos como aquellos de tendencia central, dispersión, y otras estadísticas descriptivas que traducen características de la distribución de los datos en un único número.

La **estadística inferencial** por otra parte, se ocupa de *hacer predicciones o inferencias* sobre una población a partir de una muestra de datos. Utiliza métodos probabilísticos para *determinar la fiabilidad de las conclusiones* extraídas y para *estimar parámetros poblacionales*. Las herramientas de la estadística inferencial incluyen intervalos de confianza, pruebas de hipótesis, y modelos de regresión, clasificación, entre otros.

### **Relación entre la Estadística Descriptiva y la Teoría de la Probabilidad**

La estadística descriptiva y la teoría de la probabilidad están estrechamente relacionadas. La teoría de la probabilidad proporciona una base matemática para describir la incertidumbre y el comportamiento aleatorio, lo cual es fundamental para la estadística inferencial. Mientras que la estadística descriptiva se centra en describir los datos observados, la teoría de la probabilidad permite modelar la incertidumbre inherente a los datos y hacer predicciones sobre eventos futuros o no observados.

### **Conceptos Fundamentales de la Estadística Descriptiva**

- **Población y Muestra**: 
  - **Población**: Conjunto completo de todos los elementos que se desea estudiar.
  - **Muestra**: Subconjunto de la población seleccionado para el estudio.

- **Parámetros y Estadísticos**:
  - **Parámetro**: Medida descriptiva calculada a partir de toda la población (por ejemplo, la media poblacional).
  - **Estadístico**: Medida descriptiva calculada a partir de una muestra (por ejemplo, la media muestral).

- **Distribución de Frecuencia**: Organización de datos que muestra la frecuencia (número de veces) de ocurrencia de cada valor o rango de valores en un conjunto de datos.

- **Medidas de Resumen**: Métricas numéricas que compactan propiedades de la distribución de frecuencia y evalúan diferentes aspectos de la mísma.
  - **Medidas de Tendencia Central**: Valores que representan un punto central o típico en un conjunto de datos. Estas medidas ayudan a identificar la ubicación central de los datos y proporcionan una indicación del "centro" de la distribución. Entre ellas encontramos la Media, mediana y moda.
  - **Medidas de Dispersión**: Grado de variabilidad o dispersión de los datos en relación con una medida de tendencia central. Estas medidas ayudan a entender la dispersión de los datos y a identificar la consistencia de los mismos. Entre ellas encontramos el Rango, varianza y desviación estándar.
  - **Medidas de Posición Relativa**: Las medidas de posición relativa indican la posición de un valor en relación con el resto del conjunto de datos. Estas medidas incluyen percentiles, cuartiles y deciles.
  - **Medidas de Forma**: Las medidas de forma describen la forma de la distribución de los datos mediante aspectos tales como la Asimetría y curtosis.

**<h2 id="section2" style="color: #007ACC;">2. Tipos de Variables</h2>**

---
Tipos de Variables dentro de la Estadística

---

En estadística, las **variables** son características o atributos que pueden tomar diferentes valores. Las variables en términos experimentales, constitullen las características del fenómeno de estudio cuyos valores son obtenidos mediante la medición de una propiedad específica del objeto, básicamente nos permiten medir y categorizar la información proveniente de un sistema. Los tipos de variables se clasifican en función de la *naturaleza de los datos y el nivel de medición*. A continuación, se presenta una descripción detallada de los tipos de variables más comunes en estadística:

**1. Variables Cualitativas (Categoricas)**

Las variables cualitativas, también conocidas como variables categóricas, describen **cualidades o atributos no numéricos** de un objeto o fenómeno. Estas variables no implican un orden o cantidad, sino una clasificación en diferentes categorías. Pueden dividirse en dos subtipos:

  - **1.1 Nominales**:
    Las variables nominales son aquellas que representan categorías **sin un orden intrínseco**. Los datos nominales solo indican la presencia o ausencia de una cualidad y no permiten establecer ninguna jerarquía entre las categorías.

    Ejemplos:
        - **Color de ojos**: azul, verde, marrón.
        - **Género**: masculino, femenino, no binario.
        - **Tipo de vehículo**: coche, bicicleta, motocicleta.

  - **1.2 Ordinales**:
    Las variables ordinales son similares a las nominales, pero con una diferencia importante: **las categorías tienen un orden o jerarquía natural**. Aunque las categorías pueden ser ordenadas, las diferencias entre ellas no son cuantificables ni uniformes.

    Ejemplos:
        - **Nivel educativo**: primaria, secundaria, bachillerato, licenciatura, maestría, doctorado.
        - **Nivel de satisfacción**: muy insatisfecho, insatisfecho, neutral, satisfecho, muy satisfecho.
        - **Rango militar**: soldado, cabo, sargento, teniente, capitán.

**2. Variables Cuantitativas (Numéricas)**

Las variables cuantitativas representan **cantidades numéricas** y permiten realizar operaciones aritméticas. Estas variables pueden ser medidas y ordenadas, y se dividen en dos subtipos:

  - **2.1 Discretas**:
    Las variables discretas son aquellas que toman **valores numéricos específicos y finitos**. Generalmente, representan conteos o cantidades enteras y no pueden tomar valores intermedios.

    Ejemplos:
      - **Número de hijos en una familia**: 0, 1, 2, 3, ...
      - **Número de coches en un aparcamiento**: 10, 20, 30, ...
      - **Número de estudiantes en una clase**: 25, 26, 27, ...

  - **2.2 Continuas**:
    Las variables continuas pueden tomar cualquier valor dentro de un intervalo en la línea real. Esto significa que pueden asumir un número infinito de posibles valores, incluyendo fracciones y decimales. Las variables continuas se miden, no se cuentan, y pueden representar magnitudes físicas.

    Ejemplos:
      - **Altura de una persona**: 170.5 cm, 180.2 cm, ...
      - **Peso de un objeto**: 55.3 kg, 72.1 kg, ...
      - **Tiempo de respuesta en una prueba**: 2.34 segundos, 3.56 segundos, ...

**<h2 id="section3" style="color: #007ACC;">3. Descripción de Distribuciones</h2>**

---
Descripción de Distribuciones de datos

---

La **descripción de distribuciones** es un componente esencial en estadística descriptiva, ya que se enfoca en resumir y representar cómo se distribuyen los valores de una variable en un conjunto de datos. Este análisis permite identificar patrones y características importantes que definen la naturaleza de los datos, ayudando a los analistas y tomadores de decisiones a comprender mejor la información y a realizar inferencias adecuadas.

### **¿Para qué sirve?**

La descripción de distribuciones permite:

1. **Identificar patrones y tendencias**: Al observar cómo los datos se distribuyen, es posible detectar patrones como la concentración de datos en ciertas áreas o la dispersión de los mismos.

2. **Comprender la variabilidad**: Se analiza qué tan dispersos o concentrados están los datos, lo que es crucial para evaluar la consistencia y precisión de los datos recogidos.

3. **Detectar valores atípicos**: La descripción de la distribución puede revelar valores extremos que pueden ser de interés o requerir atención especial, como en el caso de errores de medición o datos excepcionales.

4. **Comparar diferentes conjuntos de datos**: Facilita la comparación entre diferentes grupos o condiciones para entender similitudes o diferencias en los datos.
    
5. **Tomar decisiones informadas**: Proporciona una base sólida para la toma de decisiones basada en datos, ya sea en el ámbito de negocios, ciencia, ingeniería, entre otros.

### **Tipos de Descripciones de Distribuciones**

En estadística descriptiva, se suelen considerar varias dimensiones para describir una distribución, entre las que se destacan:

1. **Medidas de Tendencia Central**: Estas medidas indican dónde se encuentra el "centro" de los datos. Incluyen la media, la mediana y la moda. Son útiles para entender qué valores son típicos en el conjunto de datos.

2. **Medidas de Dispersión**: Describen cómo se dispersan o distribuyen los datos alrededor de las medidas de tendencia central. Ejemplos incluyen el rango, la varianza y la desviación estándar. Ayudan a entender la variabilidad de los datos.

3. **Medidas de Posición Relativa**: Estas medidas, como los percentiles y cuartiles, ayudan a identificar la posición relativa de un valor dentro de la distribución. Son útiles para entender cómo se compara un valor específico con el resto del conjunto de datos.

4. **Medidas de Forma**: Incluyen la asimetría (skewness) y la curtosis (kurtosis), que describen la forma de la distribución. La asimetría indica si los datos están sesgados hacia un lado, mientras que la curtosis mide la "puntiagudez" de la distribución.

**<h3 id="subsection31" style="color: #003366;">3.1. Medidas de Tendencia Central</h3>**

---
Medidas de Tendencia Central 

---

Las **medidas de tendencia central** son estadísticas fundamentales en la descripción de distribuciones, ya que proporcionan una idea general sobre el "centro" o punto de referencia principal de un conjunto de datos. Estas medidas ayudan a identificar el **valor típico o promedio en un conjunto de datos** y son esenciales para resumir y comparar diferentes distribuciones. A continuación, se presentan las principales medidas de tendencia central y sus características:

1. **Media Aritmética**: Comúnmente conocida como "media", es la medida de tendencia central más utilizada. Se calcula sumando todos los valores en un conjunto de datos y dividiendo el resultado entre el número total de valores. Constituye el "centro de masa" de la distribución de los datos y tiene una equivalencia directa con el valor esperado de la distribución.

$$
\mu = \frac{\sum_{i=1}^{n} x_i}{n} 
$$

$$
\mu = E[X] = \sum_{i=1}^{n} x_i p_i
$$

Dentro de sus más comúnes características encontramos:

- **Representativa**: Ofrece una visión general del conjunto de datos.

- **Sensibilidad a Valores Extremos**: Puede verse muy afectada por valores atípicos o extremos, ya que cada valor contribuye directamente al cálculo de ésta métrica.


2. **Mediana**: Corresponde al valor que se encuentra en el **centro de un conjunto de datos ordenado de menor a mayor**. Si el número de observaciones es impar, la mediana es el valor del medio. Si el número de observaciones es par, la mediana es el promedio de los dos valores centrales. Esto quiere decir que la mediana representa el valor en donde se encuentran el 50% de los datos acumulados dentro de la distribución.

$$
\text{Mediana} = \begin{cases}
x_{\frac{n+1}{2}} & \text{si } n \text{ es impar} \\
\frac{x_{\frac{n}{2}} + x_{\frac{n}{2}+1}}{2} & \text{si } n \text{ es par}
\end{cases}
$$

Dentro de sus más comúnes características encontramos:

- **Resistente a Valores Extremos**: No se ve afectada por valores atípicos ya que solo considera el valor central.

- **Útil para Datos Asimétricos**: Proporciona una mejor representación del centro en distribuciones sesgadas.


3. **Moda**: Es el valor o los valores que **ocurren con mayor frecuencia** en un conjunto de datos. Un conjunto de datos puede tener una única moda (unimodal), más de una moda (multimodal), o no tener moda si todos los valores ocurren con la misma frecuencia.

$$
\text{Moda} = \arg \max_{x} f(x)
$$

Dentro de sus más comúnes características encontramos:

- **Utilidad para Datos Categóricos**: Es especialmente útil para datos cualitativos donde se desea identificar la categoría más frecuente.

- **No siempre única**: Un conjunto de datos puede tener más de una moda o ninguna moda si todos los valores aparecen con la misma frecuencia.

````python
# Función para calcular la media
def calcular_media(datos):
    return sum(datos) / len(datos)

# Función para calcular la mediana
def calcular_mediana(datos):
    datos_ordenados = sorted(datos)
    n = len(datos_ordenados)
    mitad = n // 2
    if n % 2 == 0:
        # Si el número de datos es par, calcular el promedio de los dos valores centrales
        return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2
    else:
        # Si el número de datos es impar, devolver el valor central
        return datos_ordenados[mitad]

# Función para calcular la moda
def calcular_moda(datos):
    conteo = Counter(datos)
    max_frecuencia = max(conteo.values())
    modas = [key for key, value in conteo.items() if value == max_frecuencia]
    return modas
````

````python
# Calculamos las medidas de tendencia central para cada variable

# Calculamos las métricas para variables numéricas
resultados = pd.DataFrame(columns=['Variable', 'Medias', 'Medianas','Modas'])
medias = []
medianas = []
modas = []
numericas = df_turismo.drop(columns=['UserID']).select_dtypes(exclude=['object']).columns

for variable in numericas:
    datos = df_turismo[variable]
    medias.append(calcular_media(datos))
    medianas.append(calcular_mediana(datos))
    modas.append(calcular_moda(datos))

resultados['Variable'] = numericas
resultados['Medias'] = medias
resultados['Medianas'] = medianas
resultados['Modas'] = modas

# Imprimimos los resultados
resultados
````

**<h3 id="subsection32" style="color: #003366;">3.2. Medidas de Dispersión</h3>**

---
Medidas de Disperción y la cohesión de los datos 

---

Las **medidas de dispersión** son estadísticas que describen la **variabilidad o dispersión de los valores en un conjunto de datos**. Mientras que las medidas de tendencia central (media, mediana, moda) nos indican el valor típico o central de los datos, las medidas de dispersión nos informan sobre cuán dispersos o concentrados están los datos en torno a ese valor central. Entender la dispersión es crucial para interpretar correctamente los datos y para realizar comparaciones entre diferentes conjuntos de datos ya que nos dan una perspectiva a cerca de que tanta cohesión exíste en los datos y dan un indicio que que tanta información puede contener una variable. Dentro de las medidas de dispersión encontramos:

1. **Rango**: Es la diferencia entre el valor máximo y el valor mínimo en un conjunto de datos, constituye la medida de disperción más básica y es el intervalo numérico dentro del cuál una variable se observa. Nos da un indicio de la escala y el ancho de variabilidad de una variable. No obstante, al ser una medida relacionada con la amplitud de la variable, es la medida más sensible a los valores atípicos dentro de todo el conjunto de estadísticos de una variable.

   $$
   \text{Rango} = x_{max} - x_{min} 
   $$


2. **Varianza**: La varianza mide la dispersión de los datos en relación con la media. Es el promedio de las diferencias al cuadrado entre cada valor y la media de la variable. La varianza nos da una medida cuadrática de la dispersión, haciendo que los valores sean más grandes cuando los datos están más dispersos o lejanos del valor central. La .Siendo $x_i$ el valor i-ésimo posible para la variable y  $\mu$ el promedio de la mísma para un total de $n$ observaciones, la varianza viene dada por:
   
   $$
    \sigma^2 = \frac{\sum_i (x_i - \mu)^2}{N} 
   $$

3. **Desviación Estándar**: La desviación estándar es la raíz cuadrada de la varianza. Proporciona una medida de dispersión en las mismas unidades que los datos originales y suele ser más usada por ésta razón, para determinar proporcionalidades entre los datos como por ejemplo, en el caso de la detección de outliers o valores atípicos, un valor $x_i$ es considerado como anómalo si está entre dos a tres desviaciones standard por encima o por debajo de la media.
   
   $$
    \sigma = \sqrt{\frac{\sum_i (x_i - \mu)^2}{N}} 
   $$

4. **Cuartiles**: Los cuartiles son medidas estadísticas que dividen un conjunto de datos ordenados en cuatro partes iguales. Se utilizan para describir la distribución de los datos y proporcionar información sobre la dispersión y la posición de los valores dentro del conjunto de datos. Los tres cuartiles principales son:

$$
Q1 = x_{\frac{n+1}{4}}
$$

$$
Q2 = \begin{cases}
x_{\frac{N+1}{2}} & \text{si } N \text{ es impar} \\
\frac{x_{\frac{N}{2}} + x_{\frac{N}{2}+1}}{2} & \text{si } N \text{ es par}
\end{cases}
$$

$$
Q3 = x_{\frac{3(N+1)}{4}}
$$

5. **Rango Intercuartílico (IQR)**: El rango intercuartílico mide la dispersión en el rango central de los datos, es decir, entre el primer cuartil (Q1) y el tercer cuartil (Q3) los cuáles representan el valor de la variable $X$ hasta donde el 25% y 75% de la cantidad de observaciones totales respectivamente se encuentran. Corresponde una medida menos sensible a los valores atípicos y nos permíte tener un criterio para identificarlos.
   
   $$
     \text{IQR} = Q3 - Q1 
   $$

6. **El coeficiente de variación (CV)**: Es una medida estadística que se utiliza para cuantificar la variabilidad relativa de una variable en relación con su media. Se calcula como la desviación estándar de la variable dividida por su media, y se expresa como un porcentaje. El coeficiente de variación es útil cuando se desea comparar la variabilidad de diferentes conjuntos de datos que pueden tener unidades de medida diferentes o escalas distintas. Por ejemplo, si se están comparando dos distribuciones de ingresos en diferentes países, el CV permite evaluar cuál de las distribuciones es más variable en relación con su media, independientemente de las unidades de moneda utilizadas en cada país. Un coeficiente de variación bajo indica que la variabilidad de la variable es relativamente pequeña en relación con su media, mientras que un coeficiente de variación alto indica una mayor variabilidad relativa en comparación con la media. El CV se expresa como un porcentaje para facilitar su interpretación y comparación entre diferentes variables o conjuntos de datos.

$$ CV (X) = \frac{std(X)*100}{mean(X)} $$


Las medidas de dispersión pueden visualizarse de diversas maneras dentro de algúnos gráficos:

1. **Histogramas**: Los histogramas muestran la distribución de los datos y pueden dar una idea visual de la dispersión. Un histograma con barras anchas y bajas indica alta dispersión, mientras que barras estrechas y altas indican baja dispersión.

2. **Diagramas de Caja (Box Plots)**: Los diagramas de caja son útiles para visualizar el rango, la mediana, el IQR y posibles valores atípicos. La caja central muestra el rango intercuartílico, mientras que los bigotes (líneas que salen de la caja) muestran el rango general de los datos.

3. **Gráficos de Dispersión**: Los gráficos de dispersión muestran la relación entre dos variables y pueden indicar la dispersión de los datos en dos dimensiones. Los puntos dispersos indican alta variabilidad, mientras que los puntos concentrados  en alguna parte del espacio indican baja variabilidad.

Las medidas de dispersión son esenciales en muchos contextos del análisis de datos tales como:

- **Comparación de Grupos**: Ayudan a comparar la variabilidad entre diferentes grupos o poblaciones.
- **Identificación de Valores Atípicos**: Facilitan la detección de valores atípicos que pueden influir en el análisis.
- **Análisis de la Fiabilidad de la Media**: Complementan las medidas de tendencia central para proporcionar una imagen más completa de los datos.

Para empezar a aprender a calcular las métricas de disperción, primero observemos colo hallar un *percentil* general donde los cuartiles Q1, Q2, Q3 y Q4 corresponden respectivamente a los percentiles 25%, 50%, 75%, y 100%. Los percentiles son también denominados **métricas de posición relativa**  porque indican la posición de un valor en relación con los otros valores en un conjunto de datos. Para calcular un percentil de manera manual, usamos la siguiente expresión:

$$
Indice \left(Q_k\right) = \left( k \times \frac{N + 1}{100} \right) - 1 
$$

Donde:
- $Q_k $ es el valor del percentil $ k $.
- $ k $ es el percentil que deseas calcular.
- $ N $ es el número de datos en el conjunto de datos.
- Si el resultado no es un número entero, interpolamos entre los dos valores más cercanos.



````python
# Función para calcular manualmente los percentiles

def calcular_percentil(datos, k):

    datos_ordenados = sorted(datos)
    N = len(datos_ordenados)
    
    # Índice para el percentil
    posicion = (k / 100) * (N + 1) - 1
    
    if posicion < 0:
        return datos_ordenados[0]
    elif posicion >= N - 1:
        return datos_ordenados[-1]
    
    # Índices inferiores y superiores
    posicion_inferior = int(np.floor(posicion))
    posicion_superior = int(np.ceil(posicion))
    
    # Valores en las posiciones inferiores y superiores
    valor_inferior = datos_ordenados[posicion_inferior]
    valor_superior = datos_ordenados[posicion_superior]
    
    # Interpolación lineal si la posición no es un entero
    if posicion_inferior == posicion_superior:
        return valor_inferior
    else:
        fraccion = posicion - posicion_inferior
        return valor_inferior + fraccion * (valor_superior - valor_inferior)

# Llamamos la función
variable = "total_likes_on_outofstation_checkin_received"
datos = df_turismo[variable]
percentil_25 = calcular_percentil(datos, 25)
percentil_50 = calcular_percentil(datos, 50)  # Mediana
percentil_75 = calcular_percentil(datos, 75)

# Verificamos el uso de nuestra función mediante el cálculo de los percentiles usando pandas
percentil_25_pandas = datos.quantile(0.24)
percentil_50_pandas = datos.quantile(0.50) # Mediana
percentil_75_pandas = datos.quantile(0.75)

print('*'*36)
print(f"Percentil 25: {percentil_25}")
print(f"Percentil 50 (Mediana): {percentil_50}")
print(f"Percentil 75: {percentil_75}")
print('*'*36)
print(f"Percentil 25 por pandas: {percentil_25_pandas}")
print(f"Percentil 50 (Mediana) por pandas: {percentil_50_pandas}")
print(f"Percentil 75 por pandas: {percentil_75_pandas}")
print('*'*36)
````

Una vez hemos comprendido como los percentiles se calculan,  podemos proceder a calcular las demás métricas de dipersión de manera manual. En adición a esto, calcularemos el límite superior e inferior bajo los cuales consideramos una observación de los datos como dato atípico mediante el uso de las siguientes relaciones:

$$
x_{\text{sup}} = Q_3 + 1.5 \cdot \text{IQR}
$$

$$
x_{\text{inf}} = Q_1 - 1.5 \cdot \text{IQR}
$$

Note a demás, que las métricas de dispersión como la varianza y la desviación estandar, pueden ser calciladas mediante pandas usando los comandos `data.var()` y `data.std()` respectivamente, mientras que los percentiles se calculan usando las sentencias `data.quantile(p)` donde `p` es el valor de cero a uno que representa el percentil que deseamos calcular.

````python
# Cálculo de las medidas de dispersión de manera manual

def calcular_medidas_dispersion(df):
    dispersion_dict = {}

    for column in df.select_dtypes(include=[np.number]).columns:
        datos = df[column].dropna()
        n = len(datos)
        media = datos.mean()
        # Métricas de dispersión
        varianza = sum((x - media) ** 2 for x in datos) / (n - 1) 
        desviacion_estandar = varianza ** 0.5
        Q1 = datos.quantile(0.25)
        Q2 = datos.quantile(0.50)
        Q3 = datos.quantile(0.75)
        IQR = Q3 - Q1
        limite_superior_outliers = Q3 + 1.5*IQR
        limite_inferior_inferior = Q1 - 1.5*IQR
        rango = datos.max() - datos.min()
        coeficiente_variacion = (desviacion_estandar / media) * 100
        
        dispersion_dict[column] = {
            'Media': media,
            'Varianza': varianza,
            'Desviación Estándar': desviacion_estandar,
            'Q1 (Cuartil 1)': Q1,
            'Q2 (Cuartil 2)': Q2,
            'Q3 (Cuartil 3)': Q3,
            'IQR (Rango Intercuartílico)': IQR,
            'Límite Inferior (IQE)': limite_inferior_inferior,
            'Límite Superior (IQE)': limite_superior_outliers,
            'Rango': rango,
            'Coeficiente de Variación (%)': coeficiente_variacion
        }

    return pd.DataFrame(dispersion_dict)

# Calcular las medidas de dispersión
medidas_dispersion = calcular_medidas_dispersion(df_turismo.drop(columns=['UserID']))
medidas_dispersion
````

**<h3 id="subsection33" style="color: #003366;">3.3. Medidas de Forma</h3>**

---
Medidas de Forma

---

Las **medidas de forma** son estadísticas que describen **la distribución de los datos en torno a su medida de tendencia central** (como la media o la mediana) y complementan las medidas de centralidad (media, mediana) y dispersión (varianza, desviación estándar) al dar un indicativo cuantitativo de propiedades de forma tales como la **concentración de los datos** y el **sesgo de los datos**. Estas medidas ayudan a entender no solo la ubicación central de los datos, sino también cómo se distribuyen alrededor de esa ubicación. Los tipos de métricas de forma son:


1. **Asimetría (Sesgo o skewness)**: La **asimetría**, sesgo o skewness mide la simetría de la distribución de los datos en torno al valor central. En otras palabras, indica si la distribución está *sesgada hacia la derecha (asimetría positiva)* o hacia *la izquierda (asimetría negativa)*.

   - **Asimetría Positiva**: La cola derecha es más larga o más gruesa que la cola izquierda. Los valores extremos tienden a estar en el extremo superior.
   - **Asimetría Negativa**: La cola izquierda es más larga o más gruesa que la cola derecha. Los valores extremos tienden a estar en el extremo inferior.

   La asimetría **poblacional** calculada sobre una distribución de población viene dada por:

   $$
   \gamma = \frac{\frac{1}{N} \sum_{i=1}^N (x_i - \mu)^3}{\left(\frac{1}{N} \sum_{i=1}^N (x_i - \mu)^2\right)^{3/2}} 
   $$

   Lo cuál también puede rescribirse como:

   $$
   \gamma = E \left[ \left( \frac{X - \mu}{\sigma} \right)^3 \right] = \frac{\mu_3}{\sigma^3}
   $$

   Por otra parte, la asimetría calculada sobre una **muestra de la población** de tamaño $n$ se ajusta para ser una estimación insesgada del sesgo poblacional. Si $s$ es la varianza muestral y $\bar{x}$ es la media muestral, para una distribución de datos de una fracción de la población el coeficiente de asimetría se calcula como:

   $$
   \gamma_s = \frac{n}{(n-1)(n-2)} \sum_{i=1}^n \left(\frac{x_i - \bar{x}}{ s }\right)^3
   $$

   Note que la **asimetría muestral** utiliza el **sesgo estandarizado** (es decir, en términos de desviación estándar muestral). En esta forma, la fórmula tiene en cuenta *la corrección por el tamaño muestral* para proporcionar una estimación insesgada del sesgo poblacional. La expresión $\frac{N}{(N-1)(N-2)}$ se utiliza para *ajustar la fórmula muestral del sesgo y compensar la estimación sesgada del sesgo en muestras finitas*. Esto se debe a que el sesgo calculado directamente de una muestra no es completamente insesgado, y este ajuste es necesario para obtener una estimación más precisa del sesgo poblacional.

   Para interpretar correctamente la asimetría de una distribución, tenemos en cuenta los siguientes aspectos:
   
   - Un valor de asimetría cercano a 0 indica una distribución aproximadamente simétrica.
   - Valores positivos indican asimetría hacia la derecha.
   - Valores negativos indican asimetría hacia la izquierda.


2. **Curtosis**: La curtosis mide la **"altitud" y "anchura"** de las colas de la distribución en comparación con una distribución normal. Es una *medida de la forma de las colas de la distribución*.

   - **Curtosis Alta**: La distribución tiene colas más pesadas (más pronunciadas) y un pico más alto que la distribución normal (más valores extremos).
   - **Curtosis Baja**: La distribución tiene colas más ligeras (menos pronunciadas) y un pico más bajo que la distribución normal (menos valores extremos).

   La curtosis **poblacional** calculada sobre una distribución de población viene dada por:

   $$
   \beta = \frac{\frac{1}{N} \sum_{i=1}^N (x_i - \bar{x})^4}{\left(\frac{1}{N} \sum_{i=1}^N (x_i - \mu)^2\right)^2} 
   $$

   Lo cuál también puede rescribirse como:

   $$
   \beta = E \left[ \left( \frac{X - \mu}{\sigma} \right)^4 \right] = \frac{\mu_4}{\sigma^4}
   $$

   No obstante, en la distribución normal se tiene que  $\mu_4 = 3\sigma^4$ donde $\mu_4$ es el **cuarto momento centrado** con respecto a la media, no obstante si nuestro punto de referencia es la curva gaussiana, para que $\beta$ sea cero necesitamos substaer el coeficiente $3$ de nuestra previa definición de la curtosis. Así, una denominación más utilizada es el **exceso de curtosis** dada simplemente por:

   $$
   \beta = \frac{\mu_4}{\sigma^4} - 3
   $$ 
   
   La **curtosis muestral** ajusta la fórmula para reflejar las estimaciones basadas en una muestra finita en lugar de la población completa. La expresión para la curtosis muestral es calculada con base al tamaño de la muestra $n$, su media $\bar{x}$ y su varianza muestral $s$.

   $$
   \beta_s = \frac{n(n+1)}{(n-1)(n-2)(n-3)} \sum_{i=1}^n \left(\frac{x_i - \mu}{s}\right)^4 - \frac{3(n-1)^2}{(n-2)(n-3)}
   $$

   

   Para interpretar correctamente la asimetría de una distribución, tenemos en cuenta los siguientes aspectos:

   - Una curtosis de 0 (después de ajustar) indica una distribución que se comporta como una distribución normal (mesocúrtica).
   - Curtosis positiva indica colas más pesadas y un pico más alto (leptocúrtica).
   - Curtosis negativa indica colas más ligeras y un pico más bajo (platicúrtica).


### **Relación con Medidas de Centralidad y Dispersión: Momentos Centrados**

Los **momentos centrados** son una extensión de los momentos tradicionales en estadística que se utilizan para describir propiedades de la distribución de una variable aleatoria. Los momentos centrados se calculan en relación con la media de la distribución y proporcionan información más detallada sobre la forma de la distribución alrededor de la media. 

Los momentos tradicionales de una distribución se definen en relación con el origen o con un punto fijo (que podría ser 0). Los momentos de una variable aleatoria $X$ se definen como:

$$
M_k = E(X^k)
$$

Donde:
- $E(X^k)$ es la esperanza matemática del k-ésimo poder de $ X $.
- $k$ es el orden del momento.

Los momentos centrados se calculan en relación con la media de la distribución. Para una variable aleatoria $X$ con media $\mu$, donde el k-ésimo momento centrado se define como:

$$
\mu_k' = E[(X - \mu)^k] 
$$

En donde la información que proveen los momentos tradicionales y los momentos centrados difieren de la siguiente manera:

  - **Momentos Tradicionales**: Proporcionan información general sobre la magnitud de los valores, pero no ajustan por la ubicación central (media) de la distribución.

  - **Momentos Centrados**: Proporcionan información sobre la forma y dispersión de la distribución alrededor de la media, permitiendo medir características como la varianza, el sesgo y la curtosis.

En general, podemos observar las siguientes definiciones derivadas a partir de los momentos centrados:

- **Momento de Orden 1 (Media)**: Describe la ubicación central de la distribución. El momento centrado de orden uno es igual a cero.
- **Momento centrado de Orden 2 (Varianza)**: Describe la dispersión o la extensión de los datos respecto a la media.
- **Momento centrado de Orden 3 (Sesgo o Skewness)**: Describe la asimetría de la distribución.
- **Momento centrado de Orden 4 (Curtosis)**: Describe la forma de las colas de la distribución, es decir, la concentración de datos en las colas.

````python
# Cálculo manual de la asimetría de una población

def calcular_asimetria(df, columna):
    x = df[columna]
    n = len(x)
    media = np.mean(x)
    desviacion_std = np.std(x, ddof=0)  # Para población, ddof=0
    momento_tercero = np.mean((x - media) ** 3)
    
    asimetria = momento_tercero / (desviacion_std ** 3)
    return asimetria

# Cálculo manual de la curtosis de una población

def calcular_curtosis(df, columna):
    x = df[columna]
    n = len(x)
    media = np.mean(x)
    desviacion_std = np.std(x, ddof=0)  # Para población, ddof=0
    momento_cuarto = np.mean((x - media) ** 4)
    
    curtosis = momento_cuarto / (desviacion_std ** 4) - 3
    return curtosis

# Calculamos los indicadores de forma de nuestro dataset
formas = {}
curtosis = []
asimetrias = []

for variable in df_turismo.drop(columns=['UserID']).select_dtypes(exclude=['object']).columns:
    asimetrias.append(calcular_asimetria(df_turismo, variable))
    curtosis.append(calcular_curtosis(df_turismo, variable))
    
formas['Variable'] = df_turismo.drop(columns=['UserID']).select_dtypes(exclude=['object']).columns
formas['Asimetrias'] = asimetrias 
formas['Curtosis'] = curtosis
formas = pd.DataFrame(formas) 
formas
````

**<h3 id="subsection41" style="color: #003366;">4.1. Histogramas</h3>**

---
Histogramas 

---

Un histograma es una representación gráfica de la distribución de un conjunto de datos. Se utiliza para resumir y visualizar la distribución de una variable numérica dividiendo los datos en intervalos o **bins** y contando el número de observaciones en cada bin. Esto proporciona una vista rápida de la forma, dispersión y centralidad de los datos.

La selección del ancho de bin (o el número de bins) es crucial para la representación efectiva de un histograma ya que el ancho determinará en buena manera la cantidad de información relevante detectada por el gráfico. Por ejemplo, un número muy bajo de bins puede distorsionar la tendencia natural de agrupamiento de los datos mientras que una cantidad excesiva de bins es incapaz de generar una tendencia. Así pues, piense en el ancho y número de bins como la resolución del análisis gráfico de manera tal que un bin demasiado ancho puede ocultar detalles importantes de la distribución, mientras que un bin demasiado estrecho puede introducir ruido y hacer que la distribución parezca más errática de lo que realmente es.

Existen varios métodos para seleccionar el ancho de bin:

1. **Regla de Simple de Rango**: Corresponde al cálculo más elemental del ancho de bin donde dicha amplitud es uniforme, siendo $k$ el ancho de cada bin y $b$ el número de bins deseados, se tiene que:
   $$
   k = \frac{Rango}{b}
   $$

2. **Regla de Sturges**: La regla de Sturges es útil para captar un ancho de bin adecuado para datos aproximadamente o normalmente distribuidos o con cierta tendencia exponencial, no obstante, posee la desventaja de no ser muy adecuado para conjuntos de datos muy grandes.
   $$
   k = \lceil \log_2(N) + 1 \rceil
   $$

3. **Regla de Scott**: Adapta el ancho de los bins a la variabilidad de los datos y tiende a ser más flexible que la regla de Sturges o la regla del rango simple. Posee la desventaja de no ser adecuado para las distribuciones con una muy alta variabilidad o un sesgo demasiado produnciado. Siendo $ h $ es el ancho de bin, $ \sigma $ es la desviación estándar de los datos y $ n $ es el número de observaciones, la regla de Scott viene dada por:
   $$
   h = \frac{3.5 \sigma}{N^{1/3}}
   $$
   

4. **Regla de Freedman-Diaconis**: Emplea el rango intercuartílico para calcular el ancho del bin, lo que puede ser más robusto ante valores atípicos. Posee la desventaja de ser menos eficiente con conjuntos de datos pequeños. Donde $ h $ es el ancho de bin y $ IQR $ es el rango intercuartílico de los datos, la regla de Freedman-Diaconis viene dada por.
   $$
   h = 2 \frac{IQR}{N^{1/3}}
   $$
   

````python
# Creación de un histograma por diferentes reglas de selección de bins

class Histograma:
    def __init__(self, criterio='sturges', datos=None):
        self.criterio = criterio
        self.datos = datos
        self.num_bins = None
        self.ancho_bins = None
        
    def seleccionar_bins(self, n_bins=None):
        # Verificar que los datos no sean None
        if self.datos is None:
            raise ValueError("Los datos no pueden ser None. Proporcione un conjunto de datos.")
        
        n = len(self.datos)
        rango_datos = np.max(self.datos) - np.min(self.datos)
        if n_bins==None:
            n_bins = n ** (1 / 3)
        
        if self.criterio == 'rango':
            # Regla del Rango
            self.ancho_bins = rango_datos / (n_bins)
            self.num_bins = int(np.ceil(rango_datos / self.ancho_bins))
            
        elif self.criterio == 'sturges':
            # Regla de Sturges
            self.num_bins = int(np.ceil(np.log2(n) + 1))
            self.ancho_bins = rango_datos / self.num_bins
            
        elif self.criterio == 'scott':
            # Regla de Scott
            self.ancho_bins = 3.5 * np.std(self.datos) / n**(1/3)
            self.num_bins = int(np.ceil(rango_datos / self.ancho_bins))
            
        elif self.criterio == 'freedman-diaconis':
            # Regla de Freedman-Diaconis
            IQR = np.percentile(self.datos, 75) - np.percentile(self.datos, 25)
            self.ancho_bins = 2 * IQR / n**(1/3)
            self.num_bins = int(np.ceil(rango_datos / self.ancho_bins))
            
        else:
            raise ValueError("Criterio no reconocido. Use 'rango', 'sturges', 'scott' o 'freedman-diaconis'.")
            
    def crear_histograma(self):
        if self.num_bins is None or self.ancho_bins is None:
            self.seleccionar_bins()
        
        # Crear el histograma con los bins calculados
        plt.figure(figsize=(6, 4))
        plt.hist(self.datos, bins=self.num_bins, edgecolor='black')
        plt.xlabel('Valor')
        plt.ylabel('Frecuencia')
        plt.title(f'Histograma de los datos usando la regla de {self.criterio.capitalize()}')
        plt.grid(linestyle='--', alpha=0.7)
        plt.show()

# Ejemplo de uso
datos = np.random.normal(loc=0, scale=1, size=1000)

# Crear histogramas usando diferentes criterios
criterios = ['rango', 'sturges', 'scott', 'freedman-diaconis']
for criterio in criterios:
    hist = Histograma(criterio=criterio, datos=datos)
    hist.crear_histograma()
````

**<h3 id="subsection42" style="color: #003366;">4.2. Diagramas de Caja</h3>**

---
Diagramas de Cajas y Bigotes

---

Un diagrama de caja, también conocido como **box plot**, es una herramienta gráfica utilizada para representar la distribución de un conjunto de datos a través de sus cuartiles. Proporciona una *visualización clara de la dispersión, los valores atípicos y la asimetría de los datos*. Los diagramas de caja son especialmente útiles para comparar la distribución de varios conjuntos de datos. Dentro de los componentes de un conjunto de cajas y bigotes encontramos:

1. **Caja (Box)**:
   - Representa el rango intercuartílico (IQR), que es la distancia entre el primer cuartil (Q1) y el tercer cuartil (Q3). La caja contiene el 50% central de los datos.
   - **Límites de la Caja**:
     - **Límite Inferior**: Primer cuartil (Q1), que es el valor por debajo del cual se encuentra el 25% de los datos.
     - **Límite Superior**: Tercer cuartil (Q3), que es el valor por debajo del cual se encuentra el 75% de los datos.
   - La **línea dentro de la caja** representa la mediana (Q2), que es el valor central de los datos.

2. **Bigotes (Whiskers)**:
   - Se extienden desde los límites de la caja hasta el valor mínimo y máximo dentro de 1.5 veces el IQR desde los cuartiles. Aquellos valores que se encuentren más alla de los límites inferiores y superiores que constituyen los bigotes del plot, se denominan "valores atípicos" o "outliers".

3. **Puntos Atípicos**:
   - Valores fuera del rango definido por los bigotes. Se representan como puntos individuales en el gráfico.

Para generar un diagrama de cajas y bigotes tenemos que seguir los siguientes pasos:

1. **Calcular los Cuartiles**: Se calculan el primer cuartil (Q1), la mediana (Q2) y el tercer cuartil (Q3).

2. **Calcular el Rango Intercuartílico (IQR)** como $IQR = Q3 - Q1$. 

3. **Determinar los Bigotes**:
   - Límite Inferior de los Bigotes: $Q1 - 1.5 \times IQR$
   - Límite Superior de los Bigotes: $Q3 + 1.5 \times IQR$

4. **Graficar el Diagrama de Caja**:
   - Dibuja la caja desde Q1 hasta Q3 con una línea en Q2.
   - Agrega los bigotes desde los límites de la caja hasta los valores mínimos y máximos dentro del rango de los bigotes.
   - Marca los puntos atípicos fuera de los bigotes.

El rango intercuartílico IQR representa una medida más robusta de la dispersión de los datos diferencia de la desviación estándar, ya que no se ve afectado por los valores extremadamente grandes o pequeños que pueden influir en el cálculo de la desviación estándar. La regla de $1.5 \times \text{IQR}$ define un rango alrededor de los cuartiles donde *la mayoría de los datos deberían estar ubicados si los datos siguen una distribución normal o simétrica*. Los puntos que caen fuera de este rango se consideran inusuales. Es una convención que proporciona un buen equilibrio entre la sensibilidad y la especificidad en la detección de valores atípicos. Es un valor suficientemente amplio para capturar datos que se desvían significativamente sin ser demasiado estrictos.


````python
# Generar datos de ejemplo
np.random.seed(0)
datos = np.random.normal(loc=0, scale=1, size=100)

# Crear el diagrama de caja
plt.figure(figsize=(8, 4))
sns.boxplot(data=datos)

# Personalizar el gráfico
plt.title('Diagrama de Caja')
plt.xlabel('Datos')
plt.ylabel('Valor')
plt.grid(True, linestyle='--', alpha=0.7)

# Mostrar el gráfico
plt.show()
````

**<h3 id="subsection43" style="color: #003366;">4.3. Estimación de Densidad de Kernel (KDE)</h3>**

---
Deducción de Densidades por medio de KDE

---


La **Estimación de Densidad de Kernel** (KDE, por sus siglas en inglés Kernel Density Estimation) es una técnica *no paramétrica* para estimar la función de densidad de probabilidad de una variable aleatoria. A diferencia de los histogramas, que dividen los datos en intervalos y cuentan cuántos datos caen en cada intervalo, la KDE proporciona una representación suave y continua de la distribución de los datos. Ésto permite deducir una función de distribución contínua que representa la tendencia y comportamiento natural de los datos eliminando las discontinuidades que se generan desde el histograma debido a su formulación de conteo por intervalos. 

La KDE estima la función de densidad de probabilidad de los datos mediante la suavización de los datos usando una función kernel. *Cada punto de datos contribuye a la estimación de densidad en el espacio alrededor de él, y la suma de estas contribuciones da la estimación de la densidad total*. Así pues, para determinar la densidad KDE reaizamos los siguientes pasos:

1. **Definición de un Kernel**: Un kernel es una función de suavizado que determina la forma de la contribución de cada punto de datos. Los kernels comunes incluyen la función Gaussiana (normal), la función Epanechnikov, y la función Uniforme. La función kernel tiene una forma que se superpone con las contribuciones de otros puntos de datos, y su ancho controla el grado de suavización.

2. **Definición del ancho de banda (bandwidth)**: Es un parámetro crucial en la KDE que controla el grado de suavización y que se encuentra dentro de la función kernel, donde un ancho de banda pequeño produce una estimación más detallada pero más irregular, mientras que un ancho de banda grande produce una estimación más suave pero menos detallada. 

3. **Cálculo de la Densidad**: La función de densidad de kernel se calcula sumando las contribuciones de cada punto de datos. Matemáticamente, para un conjunto de datos $x_1, x_2, \ldots, x_n$ y un kernel $ K $ con ancho de banda $ h $, la estimación de densidad $ \hat{f}(x) $ en un punto $ x $ se calcula como:

   $$
   \hat{f}(x) = \frac{1}{n h} \sum_{i=1}^n K \left( \frac{x - x_i}{h} \right)
   $$

   donde:
   - $ n $ es el número de puntos de datos.
   - $ h $ es el ancho de banda.
   - $ K $ es la función kernel.

En la Estimación de Densidad Kernel (KDE), se utilizan varias funciones Kernel para suavizar los datos y estimar la densidad. Cada función Kernel tiene propiedades diferentes que afectan la forma de la estimación de densidad. Teniendo en cuenta que el **soporte** de un kernel es el rango de valores dentro de los cuáles la función no es nula, las funciones Kernel más comunes son:

1. **Kernel Gaussiano**

La función Kernel Gaussiana es una de las más utilizadas debido a su suavidad y propiedades matemáticas. Es particularmente popular por su naturaleza continua y suave.

$$
K(x) = \frac{1}{\sqrt{2\pi}} e^{-\frac{x^2}{2}}
$$

2. **Kernel Epanechnikov**

El Kernel Epanechnikov es más compacto y tiene una forma de campana más pronunciada que el Kernel Gaussiano. Es el Kernel óptimo en términos de eficiencia de estimación, en el sentido de que minimiza el error cuadrático medio (MSE) en la estimación de densidad.

$$
K(x) = \frac{3}{4} (1 - x^2) \quad \text{para} \quad |x| \leq 1
$$
$$
K(x) = 0 \quad \text{para} \quad |x| > 1
$$

3. **Kernel Triangular**

El Kernel Triangular es lineal dentro del intervalo y decrece a cero fuera de él. Es más simple que el Kernel Epanechnikov y tiende a ser más intuitivo, aunque no es el más eficiente en términos de MSE.

$$
K(x) = 1 - |x| \quad \text{para} \quad |x| \leq 1
$$
$$
K(x) = 0 \quad \text{para} \quad |x| > 1
$$

4. **Kernel de Uniforme**

El Kernel Uniforme es una función constante dentro de un intervalo y cero fuera de él. Es el Kernel más simple. Tiene un soporte compacto y proporciona una estimación de densidad más rugosa en comparación con otros Kernels.

$$
K(x) = \frac{1}{2} \quad \text{para} \quad |x| \leq 1
$$
$$
K(x) = 0 \quad \text{para} \quad |x| > 1
$$

5. **Kernel de Parzen**

El Kernel de Parzen es similar al Kernel Uniforme pero con soporte más grande. Proporciona una estimación de densidad suave con un soporte más amplio en comparación con el Kernel Uniforme.

$$
K(x) = \left(1 - \left|x\right|\right)^3 \quad \text{para} \quad |x| \leq 1
$$
$$
K(x) = 0 \quad \text{para} \quad |x| > 1
$$

6. **Kernel de Quartic (o Epanechnikov modificado)**

Este Kernel tiene una forma similar al Epanechnikov pero con un soporte más amplio. Proporciona una estimación de densidad aún más suave con un soporte más amplio.

$$
K(x) = \frac{15}{16} (1 - x^2)^2 \quad \text{para} \quad |x| \leq 1
$$
$$
K(x) = 0 \quad \text{para} \quad |x| > 1
$$

La selección de un kernel debe tener en cuenta los siguientes aspectos, donde la calidad de la estimación será altamente dependiente del conjunto de datos:

- **Suavidad**: Los Kernels como el Gaussiano y el Epanechnikov tienden a proporcionar estimaciones más suaves y continuas.
- **Soporte**: Kernels como el Uniforme y el Triangular tienen soporte limitado, lo que puede afectar la suavidad de la estimación.
- **Eficiencia**: El Kernel Epanechnikov es considerado el más eficiente en términos de error cuadrático medio, aunque en la práctica, la elección del Kernel puede depender de la aplicación específica y la naturaleza de los datos.

````python
# Definición de los Kernels
def kernel_gaussiano(x):
    return (1/np.sqrt(2 * np.pi)) * np.exp(-0.5 * x**2)

def kernel_epanechnikov(x):
    return np.where(np.abs(x) <= 1, 3/4 * (1 - x**2), 0)

def kernel_triangular(x):
    return np.where(np.abs(x) <= 1, 1 - np.abs(x), 0)

def kernel_uniforme(x):
    return np.where(np.abs(x) <= 1, 0.5, 0)

# Valores para x
x = np.linspace(-3, 3, 1000)

# Calcular los Kernels
kde_gaussiano = kernel_gaussiano(x)
kde_epanechnikov = kernel_epanechnikov(x)
kde_triangular = kernel_triangular(x)
kde_uniforme = kernel_uniforme(x)

# Plotear los Kernels
plt.figure(figsize=(8, 4))
plt.plot(x, kde_gaussiano, label='Kernel Gaussiano')
plt.plot(x, kde_epanechnikov, label='Kernel Epanechnikov')
plt.plot(x, kde_triangular, label='Kernel Triangular')
plt.plot(x, kde_uniforme, label='Kernel Uniforme')

# Configuración del gráfico
plt.title('Funciones Kernel')
plt.xlabel('x')
plt.ylabel('Densidad')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
````

Observemos a continuación, como en especial para datos reales, la regla para determinar el ancho de bin es crucial para obtener una buena suavización de nuestros datos.

Dentro de `Scipy`, el parámetro `kde.factor` es un multiplicador que se aplica al ancho de banda calculado automáticamente por el método de estimación de densidad kernel (KDE). Este ancho de banda es crucial para determinar la suavidad de la estimación de la densidad: un ancho de banda pequeño hace que la curva KDE sea más detallada (potencialmente más ruidosa), mientras que un ancho de banda grande la hace más suave. En KDE, el ancho de banda $h$ es un parámetro de suavizado que controla el alcance de influencia de cada punto de datos sobre el resultado de la densidad estimada.  El `kde.factor` es un valor que ajusta el ancho de banda h predeterminado calculado por un método como Scott o Silverman. Se aplica como:

$$
h_{\text{final}} = \text{kde.factor} \times h_{\text{base}}
$$

Donde $h_{\text{base}}$ es el ancho de banda calculado automáticamente, y $h_{\text{final}}$ es el ancho de banda ajustado que se usará en la estimación de la densidad. El ancho de banda h en KDE es conceptualmente similar al ancho de bin en un histograma, en el sentido de que ambos determinan cuán "agrupados" o "suaves" son los datos representados. Sin embargo, a diferencia del ancho de bin fijo en un histograma, el ancho de banda en KDE permite una representación más fluida de la densidad subyacente.

- Ancho de Bin en Histogramas: Divide los datos en intervalos disjuntos.
- Ancho de Banda en KDE: En lugar de cortar los datos en segmentos disjuntos, asigna una contribución continua de cada punto de datos a la densidad estimada, difuminando la influencia de los puntos a través de un kernel suave.

El `kde.factor` y el ancho de banda base se calculan de acuerdo con las reglas específicas de cada método (Scott, Freedman-Diaconis, Silverman):


1. **Scott’s Rule:** Determina el ancho de banda base como:

$$
h_{\text{Scott}} = \sigma \cdot n^{-\frac{1}{5}}
$$

2. **Silverman’s Rule:** Ligera variación de la regla de Scott y se calcula como:

$$
h_{\text{Silverman}} = \left(\frac{4 \sigma^5}{3n}\right)^{\frac{1}{5}} = 0.9 \cdot \min\left(\sigma, \frac{IQR}{1.34}\right) \cdot n^{-\frac{1}{5}}
$$

3. **Freedman-Diaconis Rule:** Particularmente útil para distribuciones con outliers:

$$
h_{\text{Freedman}} = 2 \cdot \frac{IQR}{n^{\frac{1}{3}}}
$$


````python
# Definimos la data
variable = "total_likes_on_outofstation_checkin_received"
datos = df_turismo[variable]

# Generamos un calculador de KDE basado enScipy que acepte diferentes métodos de ancho de banda y bin
def KDE_histograma(df, variable, bin_rule):
    datos = df[variable]
    n = len(datos)
    rango_datos = np.max(datos) - np.min(datos)
    x_values = np.linspace(min(datos), max(datos), 1000)
    IQR =  np.percentile(datos, 75) - np.percentile(datos, 25)
    sigma = np.std(datos)

    if bin_rule == 'sturges':
        num_bins = int(np.ceil(np.log2(n) + 1))
        ancho_bin = rango_datos / num_bins
        bw_factor = num_bins/(4 * (n**(1/3)))
        kde = stats.gaussian_kde(datos, bw_method=bw_factor)
        
    elif (bin_rule == 'rango' or type(bin_rule)==int):
        if type(bin_rule) == str:
            num_bins = int(np.ceil(n**(1/3)))
        else:
            num_bins = bin_rule 
        ancho_bin = rango_datos / num_bins
        bw_factor = num_bins/(4 * (n**(1/3)))
        kde = stats.gaussian_kde(datos, bw_method=bw_factor)

    elif bin_rule == 'scott':
        kde = stats.gaussian_kde(datos, bw_method='scott')
        ancho_bin = 3.5 * sigma / n**(1/3)
        num_bins = int(np.ceil(rango_datos / ancho_bin))
        bw_factor = kde.factor

    elif bin_rule == 'freedman':
        ancho_bin = 2 * IQR / n**(1/3)
        num_bins = int(np.ceil(rango_datos / ancho_bin))
        bw_factor = ( 0.05 * num_bins ) / np.min([sigma, IQR, n**(1/3)])
        kde = stats.gaussian_kde(datos, bw_method=bw_factor)


    elif bin_rule == 'silverman':
        kde = stats.gaussian_kde(datos, bw_method='silverman')
        ancho_bin = np.std(datos) * (4/(3*n))**(1/5)
        num_bins = int(np.ceil(rango_datos / ancho_bin))
        bw_factor = kde.factor

    else:
        raise ValueError("Criterio no reconocido. Use 'sturges', 'scott', 'freedman' o 'silverman'.")
    
    kde_values = kde(x_values)
    print('Ancho de banda = ', bw_factor)
    print('Numero de bins = ', num_bins)
    print('Ancho de bins = ', ancho_bin)
    print('Factor de escala = ', 4 * (n**(1/3)))
    print('Desviación estandard de los datos = ', np.std(datos))
    
    # Visualización del gráfico
    plt.figure(figsize=(8,4))
    plt.hist(datos, num_bins, density=True, alpha=0.5, color='blue', edgecolor='black', label=f'Distribución de {variable}')
    plt.plot(x_values, kde_values, color='red', label=f'KDE (Ancho de banda = {np.round(kde.factor, 2)})')
    plt.title('Estimación de Densidad Kernel')
    plt.xlabel(f'X = {variable}')
    plt.ylabel('P(X)')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.show()

# Crear la instancia KDE usando el kernel gaussiano por defecto
bandwidth = ['rango','sturges', 'scott', 'freedman', 'silverman']  

for bin_rule in bandwidth:
    print(f'Distribución de probabilidad y KDE bajo la regla de bin = {bin_rule}')
    KDE_histograma(df_turismo, variable, bin_rule)
````

Gráficos de Barras y Sectores

---

Un **gráfico de barras** representa el conteo o frecuencia de datos categóricos con rectángulos, donde la longitud de cada rectángulo es proporcional al valor que representa. Matemáticamente, si una variable categórica $X$ con categorías $C_1, C_2, \ldots, C_k $ y una función de conteo $f(C_i)$ que indica la frecuencia de la categoría $C_i$, entonces la altura de la barra asociada a $C_i$ es $f(C_i)$. Note entonces que un histograma es una variante del gráfico de barras en donde la categoría representa un intervalo de valores. La utilidad de los gráficos de barras, proveniente de su simplicidad, radica en:

- **Comparación directa de valores:** Se utilizan para comparar frecuencias, proporciones o cualquier otra métrica entre diferentes categorías.
- **Claridad:** Son efectivos cuando se tienen pocas categorías y es necesario resaltar diferencias cuantitativas.
- **Distribución discreta:** Adecuados para representar distribuciones de variables discretas como lo son los histogramas.



````python
# Generación de un gráfico de barras simple
i = 2
variable = df_turismo.select_dtypes(exclude=[int,float]).columns
datos = df_turismo[variable[i]].value_counts()
categorias = datos.index
frecuencias = datos.values

# Creación del gráfico de barras
plt.figure(figsize=(8, 4))
plt.bar(categorias, frecuencias, color='skyblue', edgecolor='black')
plt.xlabel(variable[i])
plt.ylabel('Frecuencia')
plt.title(f'Gráfico de Barras para {variable[i]}')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.show()
````

De manera anáoga a los gráficos de barras, un **gráfico de sectores o pie chart** representa datos categóricos en forma de un círculo dividido en sectores o categorías. La proporción de cada sector es equivalente al porcentaje que representa dentro del total. Matemáticamente, si la suma de todas las frecuencias $f(C_i)$ es $N = \sum_{i=1}^k f(C_i)$, entonces el ángulo de cada sector correspondiente a la categoría $C_i$ es $\theta_i = \frac{f(C_i)}{N} \times 360^\circ$. Dentro de su importancia, encontramos las siguientes razones:

- **Distribución relativa:** Ideal para mostrar la proporción de partes de un todo, donde se desea resaltar la relación entre las partes y el conjunto.
- **Datos nominales:** Útiles para variables nominales donde el orden no importa, pero la proporción sí.
- **Enfasis visual:** Permiten enfatizar la contribución relativa de diferentes categorías, aunque pueden ser menos precisos para comparaciones exactas que los gráficos de barras.

No obstante, los gráficos de torta, pie chart o de sectores al ser tan sencillos y buscar dar detalles a cerca de las proporciones que conforman al conjunto de datos, tienen la desventaja de que para un número elevado de categorías se hace dificil realmente distinguir la contribución o jerarquía de las categirías, por eso, es un gráfico limitado que se usa solo cuando el número de categorías es pequeño y es más útil cuando las proporciones son contrastantes.

````python
# Generación de un gráfico de torta simple
i = 0
variable = df_turismo.select_dtypes(exclude=[int,float]).columns
datos = df_turismo[variable[i]].value_counts()
categorias = datos.index
frecuencias = datos.values

# Creación del gráfico de sectores
plt.figure(figsize=(4, 4))
plt.pie(frecuencias, labels=categorias, autopct='%1.1f%%', startangle=140)
plt.title(f'Gráfico de Torta para {variable[i]}')
plt.show()
````

Diagramas de Violín

---

El **gráfico de violín** es una técnica de visualización de datos que *combina características de un histograma y un diagrama de caja* (boxplot). Su principal objetivo es mostrar la distribución de un conjunto de datos, proporcionando información *sobre la densidad de probabilidad y la variabilidad de los datos*. Un gráfico de violín muestra la distribución de los datos a lo largo de una variable categórica o numérica, similar a cómo un histograma muestra la distribución de frecuencias. Se llama "violín" porque su forma a menudo se asemeja a un violín o una guitarra, con una forma de doble campana que refleja la densidad de los datos. Las características principales del gráfico de violín son:

- **Distribución de Densidad:** Representa la densidad de probabilidad de los datos. La forma del violín indica la forma de la distribución en su forma PDF.
- **Medidas Resumen:** Incluye la mediana y los cuartiles, similar a un boxplot, a menudo superpuestos en el gráfico.
- **Comparación de Distribuciones:** Permite comparar la distribución de varias categorías en un solo gráfico.

Dentro de los componentes de un plot de violín encontramos:

- **Ejes:** El eje vertical representa la variable numérica $X$ (distribución de datos) y el eje horizontal representa la variable categórica (grupos o categorías).
- **Densidad de Datos:** La forma del violín muestra la densidad de datos para cada categoría. Los anchos del violín representan la densidad; un ancho mayor indica una mayor densidad de datos.
- **Medidas de Tendencia Central:** Los gráficos de violín suelen mostrar líneas o puntos para la mediana y los cuartiles, brindando una visión adicional de la distribución.

Debido a que el plot de violín combina conceptos fundamentales tales como la densidad de probabilidad, la variabilidad de los datos a través de los boxplots y el análisis comparativo entre categorías, grupos o variables en un mísmo plot, es una herramienta visual increíblemente poderosa que puede ser usada para:

- **Exploración de Datos:** Ideal para explorar y comparar distribuciones de datos en diferentes categorías o variables.
- **Comparación de Grupos:** Permite comparar cómo varían las distribuciones entre diferentes grupos o categorías.
- **Visualización de Densidad:** Ayuda a entender la forma de la distribución y detectar posibles multimodalidades (distribuciones con múltiples picos).


Para ilustrar el uso de los plots de violin, a continuación se muestra la graficación de un plot de violin donde se han generado datos aleatorios para tres grupos y se ha activado la visualización de los cuartiles dentro de los PDF de las distribuciones.

En una aplicación más real, tomarémos a continuación cuatro variables numéricas de nuestro conjunto de datos de modo tal que compararemos sus distribuciones de datos normalizados por medio de la división del malor máximo encontrado en los datos en su valor absoluto. **<h3 id="subsection47" style="color: #003366;">4.7. Gráficos QQ</h3>**

---
Gráficos QQ y corroboración de normalidad

---

Dentro de la estadística inferencial, algúnos modelos tales como la regresión lineal, regresión logística, Naive Bayes Gaussiano entre otros, junto con ciertos tests estadísticos asumen una normalidad en la distribución de los datos que modelan. No obstante aún queda la pregunta a cerca de ¿cómo corroborar si en efecto un conjunto de datos está normalmente distribuido o por lo menos, cerca de ésta forma de comportamiento estadístico?.

Un **QQ-Plot** (Quantile-Quantile Plot) es una herramienta gráfica utilizada en estadística para *comparar la distribución de un conjunto de datos con una distribución teórica, como la distribución normal*. El gráfico traza los cuantiles de los datos observados frente a los cuantiles de una distribución teórica. Si los puntos en el gráfico siguen aproximadamente una línea recta, se puede inferir que los datos siguen la distribución teórica.

Para determinar si un conjunto de datos sigue una distribución normal, se puede utilizar un QQ-Plot de la siguiente manera:

1. **Generación de Cuantiles Teóricos**: Se calculan los cuantiles teóricos correspondientes a la distribución normal para cada posición en los datos.
2. **Generación de Cuantiles de los Datos**: Se ordenan los datos observados de menor a mayor y se calculan los cuantiles observados.
3. **Trazado del QQ-Plot**: Se trazan los cuantiles observados frente a los cuantiles teóricos en un scatter plot.

Si los datos siguen una distribución normal, los puntos en el QQ-Plot deberían alinearse aproximadamente en una línea recta de modo que:

- **Datos Normales**: Los puntos caen cerca de la línea diagonal.
- **Colas Pesadas (Distribución más dispersa que la normal)**: Los puntos en las colas se alejan de la línea diagonal, curvándose hacia arriba en la parte superior derecha y hacia abajo en la parte inferior izquierda.
- **Colas Livianas (Distribución menos dispersa que la normal)**: Los puntos en las colas se curvan hacia el interior, cerca del centro del gráfico.

### **Métrica Matemática para Determinar Normalidad: Prueba de Shapiro-Wilk**

Aunque el QQ-Plot es una herramienta visual, para una evaluación más formal se puede utilizar la **Prueba de Shapiro-Wilk**. Esta prueba es un test estadístico que evalúa la hipótesis nula de que una muestra proviene de una población normalmente distribuida.

- **Hipótesis Nula (H₀)**: Los datos están normalmente distribuidos.
- **Hipótesis Alternativa (H₁)**: Los datos no están normalmente distribuidos.

El resultado de la prueba es un p-valor:

- **p-valor alto (> 0.05)**: No se rechaza la hipótesis nula, lo que sugiere que los datos pueden seguir una distribución normal.
- **p-valor bajo (≤ 0.05)**: Se rechaza la hipótesis nula, sugiriendo que los datos no siguen una distribución normal.

En el ejemplo siguiente, generarémos nuestro QQ-plot y evaluaremos este test gráfico para datos normalmente generados.

````python
# Generación de un QQ-plot manual

def qqplot(data):
    # Ordenar los datos reales
    data_sorted = np.sort(data)

    # Tamaño de la muestra
    n = len(data_sorted)

    # Generar los cuantiles teóricos para la distribución normal
    quantiles_theoretical = np.array([stats.norm.ppf((i + 1 - 0.5) / n) for i in range(n)])

    # Calculamos la media y desviación estándar de los datos observados
    mean_data = np.mean(data_sorted)
    std_data = np.std(data_sorted)

    # Estandarizar los cuantiles observados
    quantiles_observed = (data_sorted - mean_data) / std_data

    # Generar el QQ-Plot
    plt.figure(figsize=(4, 4))
    plt.scatter(quantiles_theoretical, quantiles_observed, color='blue', edgecolor='black')

    # Añadir línea diagonal que representa la distribución normal teórica
    min_val = min(quantiles_theoretical.min(), quantiles_observed.min())
    max_val = max(quantiles_theoretical.max(), quantiles_observed.max())
    plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--')
    
    # Configuración del gráfico
    plt.title('QQ-Plot Manual')
    plt.xlabel('Cuantiles Teóricos')
    plt.ylabel('Cuantiles Observados')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.show()
    
# Generamos el QQ-plot   
data = np.random.normal(loc=0, scale=1, size=1000)
# Realizar la prueba de Shapiro-Wilk
shapiro_test = stats.shapiro(data)
print('*'*36)
print(f"Shapiro-Wilk Test: Statistic={shapiro_test.statistic}, p-value={shapiro_test.pvalue}")
print('*'*36)
qqplot(data)
````

**<h2 id="section5" style="color: #007ACC;">5. Distribuciones de Probabilidad Multivariable</h2>**

---
Distribuciones de Probabilidad Multivariable

---

Las distribuciones de probabilidad multivariable son extensiones de distribuciones de probabilidad univariadas a escenarios donde hay más de una variable aleatoria involucrada. Estas distribuciones permiten modelar la relación conjunta entre múltiples variables, capturando tanto la distribución individual de cada variable como las interacciones entre ellas.

Una distribución de probabilidad multivariable describe la **probabilidad conjunta** de un conjunto de variables aleatorias. Mientras que una distribución univariada trata con una sola variable aleatoria, como la altura de las personas, una distribución multivariable puede tratar con dos o más variables simultáneamente, como la altura y el peso de las personas.

Por ejemplo, en el caso de dos variables aleatorias $X$ y $Y$, la función de distribución conjunta $f_{X,Y}(x,y)$ describe la probabilidad de que $X$ tome un valor específico $x$ y $Y$ tome un valor específico $y$ al mismo tiempo.

Las distribuciones multivariables se encuentran en muchas áreas del análisis de datos, estadística y ciencia de datos. Algunos ejemplos incluyen:

- **Modelos de Regresión:** En regresión múltiple, donde se predice una variable dependiente usando varias variables independientes.
- **Análisis de Componentes Principales (PCA):** Una técnica para reducir la dimensionalidad de datos multivariables preservando la varianza máxima.
- **Finanzas:** Modelado de retornos de diferentes activos financieros simultáneamente.
- **Ciencia de Datos:** En el modelado de datos con múltiples características (features), donde las relaciones entre estas características son importantes.

Las distribuciones de probabilidad multivariable se pueden calcular usando varias técnicas como:

1. **Función de Densidad Conjunta (PDF):** Para dos variables continuas $ X $ y $ Y $, la función de densidad conjunta $ f_{X,Y}(x,y) $ se define de manera que la probabilidad de que $ X $ esté en el intervalo $ [a,b] $ y $ Y $ en el intervalo $ [c,d] $ se calcula integrando la PDF conjunta sobre estos intervalos:
   $$
   P(a \leq X \leq b, c \leq Y \leq d) = \int_{a}^{b}\int_{c}^{d} f_{X,Y}(x,y) \, dy \, dx
   $$

2. **Distribución Marginal:** Si solo estamos interesados en una de las variables, podemos derivar su distribución marginal integrando la función de densidad conjunta sobre los posibles valores de la otra variable. Por ejemplo, la distribución marginal de $ X $ es:
   $$
   f_X(x) = \int_{-\infty}^{\infty} f_{X,Y}(x,y) \, dy
   $$

3. **Distribución Condicional:** La distribución condicional de $ Y $ dado $ X = x $ se obtiene de la función de densidad conjunta y se usa para analizar la relación entre las variables:
   $$
   f_{Y|X}(y|x) = \frac{f_{X,Y}(x,y)}{f_X(x)}
   $$


Las distribuciones de probabilidad multivariable son fundamentales para modelar sistemas donde varias variables aleatorias interactúan entre sí. Se encuentran en aplicaciones desde la estadística básica hasta el análisis de datos complejo, proporcionando herramientas para entender y predecir la relación conjunta entre diferentes variables. Así pues, cualquier conjunto de datos de un sistema real corresponde a un espacio muestral multivariado que inevitablemente puede definir este tipo de distribuciones de probabilidad.

**<h3 id="subsection51" style="color: #003366;">5.1. Distribuciones Conjuntas</h3>**

---
Distribuciones de Probabilidad Conjuntas

---

Las **distribuciones de probabilidad conjuntas** describen la probabilidad de que dos (o más) variables aleatorias tomen ciertos valores simultáneamente, pasí pues son una distribución de probabilidad multivariable por definición. Estas distribuciones son fundamentales en estadística y probabilidad, ya que permiten analizar la relación entre varias variables, facilitando el estudio de su comportamiento conjunto.

### **Distribuciones Discretas**

En el caso de variables aleatorias discretas, una distribución de probabilidad conjunta se define a través de una **función de probabilidad conjunta**. Supongamos que $X$ e $Y$ son dos variables aleatorias discretas. La función de probabilidad conjunta $ P(X = x, Y = y)$ nos da la probabilidad de que $ X $ tome el valor $ x $ e $ Y $ tome el valor $ y $ al mismo tiempo.

Matemáticamente, si $ X $ puede tomar los valores $ x_1, x_2, \ldots, x_n $ y $ Y $ puede tomar los valores $ y_1, y_2, \ldots, y_m $, entonces la distribución conjunta de $ X $ e $ Y $ se puede representar mediante una matriz donde cada entrada $p_{XY}(x_i,y_j) = P(X = x_i, Y = y_j) $ indica la probabilidad conjunta correspondiente.

Para que esta distribución sea válida, debe cumplir con dos condiciones:

1. **No negatividad**: $ P(X = x_i, Y = y_j) \geq 0 $ para todos $ i $ y $ j $.
2. **Suma total igual a 1**: La suma de todas las probabilidades conjuntas debe ser igual a 1.
   
   $$
   \sum_{i=1}^{n} \sum_{j=1}^{m} P(X = x_i, Y = y_j) = 1
   $$


Para calcular una matriz de probabilidad conjunta, es necesario observar cada combinación posible entre los valores de una variable y otra, realizar un conteo de la presencia de cada combinación y dividir cada cuenta calculada por el número total de observaciones, proceso que se hace extremadamente complejo cuando exísten múltiples valores edentro cada variable y múltiples pariables. No obstante, éste cálculo es posible de realizar en Python mediante  `pd.crosstab`, métoro usado para crear una matriz de frecuencias conjunta que calcula la frecuencia de cada combinación de valores para las variables especificadas.

La función `pd.crosstab` en pandas se utiliza para crear una tabla de contingencia que muestra la frecuencia conjunta de dos o más variables. Es particularmente útil para analizar datos categóricos. La sintaxis de `pd.crosstab` es la siguiente:

```python
pd.crosstab(index, columns, values=None, aggfunc=None, rownames=None, colnames=None, dropna=True, margins=False, margins_name='All', **kwargs)
```

**Parámetros Principales**

1. **`index`**: 
   - **Descripción**: Los datos que se utilizarán para las filas de la tabla.
   - **Tipo**: Array-like (list, Series, etc.)

2. **`columns`**: 
   - **Descripción**: Los datos que se utilizarán para las columnas de la tabla.
   - **Tipo**: Array-like (list, Series, etc.)

3. **`values`** (opcional): 
   - **Descripción**: Valores que se utilizarán para llenar la tabla en lugar de contar frecuencias. Por ejemplo, para calcular la suma o el promedio.
   - **Tipo**: Array-like (list, Series, etc.)

4. **`aggfunc`** (opcional): 
   - **Descripción**: Función de agregación que se aplicará a los `values` si se proporciona. Por defecto es `np.sum`.
   - **Tipo**: Function (e.g., `np.sum`, `np.mean`)

5. **`rownames`** (opcional): 
   - **Descripción**: Etiquetas para las filas.
   - **Tipo**: List of strings

6. **`colnames`** (opcional): 
   - **Descripción**: Etiquetas para las columnas.
   - **Tipo**: List of strings

7. **`dropna`** (opcional): 
   - **Descripción**: Si se debe eliminar las filas y columnas que contienen únicamente valores `NaN`.
   - **Tipo**: Boolean (default `True`)

8. **`margins`** (opcional): 
   - **Descripción**: Si se deben agregar totales marginales (suma de filas y columnas).
   - **Tipo**: Boolean (default `False`)

9. **`margins_name`** (opcional): 
   - **Descripción**: Nombre que se usará para la fila/columna de márgenes.
   - **Tipo**: String (default `'All'`)

10. **`**kwargs`** (opcional): 
    - **Descripción**: Otros argumentos adicionales que se pueden pasar a la función.

En resumen, se tiene que:

- **`index`** y **`columns`**: Especifican las variables categóricas para las filas y columnas.
- **`values`**: (Opcional) Proporciona valores numéricos para agregación.
- **`aggfunc`**: (Opcional) Define cómo agregar los valores numéricos.
- **`margins`**: Añade totales marginales para filas y columnas.



A continuación, veremos unos ejemplos básicos de uso de éste método combinatorio para poder entender su lógica.

Una vez expuesto el funcionamiento del cálculo de probabilidades por combinaciones, generaremos una función que en efecto, calcule la matriz que representa la distribución conjunta.

````python
# Definición de una finción para generar una distribución discreta multivariable

def probabilidad_conjunta(df, variables):
    
    # Crear una tabla de frecuencias conjunta
    tabla_frecuencias = pd.crosstab(index=[df[var] for var in variables], 
                                   columns=pd.Index([1], name='count'))

    # Convertir la tabla de frecuencias a una tabla de probabilidades
    total_observaciones = tabla_frecuencias.sum().sum()
    prob_conjunta = tabla_frecuencias / total_observaciones

    return prob_conjunta

# Ejemplo de uso
# Supongamos que tenemos un DataFrame df_simple con variables categóricas 'var1', 'var2', 'var3'
df_turismo_test = pd.DataFrame({
    'var1': ['A', 'B', 'A', 'A', 'B', 'B', 'A'],
    'var2': ['X', 'Y', 'Y', 'X', 'X', 'Y', 'Y'],
    'var3': [1, 1, 2, 3, 1, 1, 2]
})

variables = ['var1', 'var2','var3']
prob_conjunta = probabilidad_conjunta(df_turismo_test, variables)

print("Probabilidades conjuntas:")
print(prob_conjunta)
````

La **probabilidad marginal** de una variable en una distribución conjunta se obtiene sumando las probabilidades conjuntas sobre todas las posibles valores de la otra(s) variable(s). Para una variable $ X $ o $Y$ en una distribución conjunta $ P(X, Y) $ se tiene que:

- **Probabilidad Marginal de $ X $**, $ P(X) $:
  $$
  P(X = x_i) = \sum_{j} P(X = x_i, Y = y_j)
  $$
  donde la suma es sobre todos los posibles valores de $ Y $.

- **Probabilidad Marginal de $ Y $**, $ P(Y) $:
  $$
  P(Y = y_j) = \sum_{i} P(X = x_i, Y = y_j)
  $$
  donde la suma es sobre todos los posibles valores de $ X $.

En el caso de variables aleatorias continuas, la distribución de probabilidad conjunta se describe mediante una **función de densidad de probabilidad conjunta** (f.d.p. conjunta). Si $X$ e  $Y$ son variables continuas, la f.d.p. conjunta se denota como  $f_{X,Y}(x,y)$.

Esta función cumple con las siguientes propiedades:

1. **No negatividad**: $f_{X,Y}(x,y) \geq 0 $ para todos los valores de $x$ e $ y $.
2. **Integración a 1**: La integral doble de la f.d.p. conjunta sobre todo el espacio de posibles valores de $ X $ e $ Y $ debe ser igual a 1.

   $$
   \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} f_{X,Y}(x,y) \,dx\,dy = 1
   $$

3. **Probabilidades conjuntas**: La probabilidad de que $X$ e $Y$ caigan en un intervalo dado se obtiene mediante la integración de la f.d.p. conjunta sobre ese intervalo:

   $$
   P(a \leq X \leq b, c \leq Y \leq d) = \int_{a}^{b} \int_{c}^{d} f_{X,Y}(x,y) \,dx\,dy
   $$

4. **Distribución Marginal:** Si solo estamos interesados en una de las variables, podemos derivar su distribución marginal integrando la función de densidad conjunta sobre los posibles valores de la otra variable. Por ejemplo, la distribución marginal de $ X $ es:
   $$
   f_X(x) = \int_{-\infty}^{\infty} f_{X,Y}(x,y) \, dy
   $$

Distribuciones de Probabilidad Condicional

---


La **distribución de probabilidad condicional** mide la probabilidad de que ocurra un evento dado que otro evento ya ha ocurrido. Matemáticamente, la probabilidad condicional de un evento $A$ dado que $B$ ha ocurrido se denota como $P(A \mid B)$ y se define como:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

En el contexto multivariado, estamos interesados en la probabilidad condicional de un subconjunto de variables dado otro subconjunto. Esto se puede aplicar tanto a variables **discretas** como **continuas**.

### **Distribuciones de Probabilidad Condicional para Variables Discretas**

Para un conjunto de variables discretas $ X_1, X_2, \dots, X_n $, la distribución de probabilidad conjunta se representa por $ P(X_1 = x_1, X_2 = x_2, \dots, X_n = x_n) $. La distribución de probabilidad condicional de un subconjunto de variables dado otro subconjunto se puede calcular utilizando la relación:

$$
P(X_1 = x_1, \dots, X_k = x_k \mid X_{k+1} = x_{k+1}, \dots, X_n = x_n) = \frac{P(X_1 = x_1, \dots, X_k = x_k, X_{k+1} = x_{k+1}, \dots, X_n = x_n)}{P(X_{k+1} = x_{k+1}, \dots, X_n = x_n)} 
$$

Para obtener la distribución de probabilidad condicional  seguimos los siguientes pasos:

1. **Obtener la distribución conjunta**: Primero se determina la distribución conjunta de todas las variables involucradas. Esta puede representarse como una tabla multidimensional.

2. **Marginalización**: Se calcula la probabilidad marginal del subconjunto de variables sobre el cual se condiciona. Esto se hace sumando las probabilidades sobre las otras variables.

3. **División**: Se divide la probabilidad conjunta por la probabilidad marginal obtenida en el paso anterior.

Por ejemplo, en el caso en el que solo tengamos dos variables $X$ y $Y$, donde $P_{XY}(x,y)$ es la probabilidad conjunta discreta de las dos variables calculable mediante la tabla de combinaciones y conteos normalizados que vimos en la sección anterior. 

$$
P_{XY}(xy) = \frac{C(x_i,y_j)}{N}
$$

donde $C(x,y)$ es el número de veces en que encontramos simultáneamente encontramos los valores $X=x$ y $Y=y$ en el conjunto de datos y $N$ el tamaño del dataset. Por otra parte, tenemos que la probabilidad marginal viene dada por:

$$
P_X(x) = \sum_{i} P(x_i) = \sum_{ij}P_{XY}(x_i,y_j)
$$

Así pues, la probabilidad condicional de una variable $Y$ dado un valor de $X$, en el caso discreto, es igual a tomar cada entrada de la matriz de  probabilidad conjunta $P_{XY}(x_i,y_j)$  y dividirla por la suma de la fila (o columna) correspondiente al valor $x_i$ (esta suma corresponde a una de as entradas de la distribución marginal de $X$) tal que: 

$$
P_X(x_i) = \sum_j P_{XY}(x_i,y_j)
$$

De esta manera, la probabilidad condicional discreta es igual a:

$$
P_{Y|X}(y_j|x_i) = \frac{P_{XY}(x_i,y_j)}{P_X(x_i)}
$$

### **Distribuciones de Probabilidad Condicional para Variables Continuas**

Para un conjunto de variables continuas $X_1, X_2, \dots, X_n $, la distribución conjunta se describe por la **densidad de probabilidad conjunta** $ f_{X_1, X_2, \dots, X_n}(x_1, x_2, \dots, x_n) $. La densidad de probabilidad condicional de un subconjunto de variables dado otro subconjunto es:

$$
f_{X_1, \dots, X_k \mid X_{k+1}, \dots, X_n}(x_1, \dots, x_k \mid x_{k+1}, \dots, x_n) = \frac{f_{X_1, \dots, X_k, X_{k+1}, \dots, X_n}(x_1, \dots, x_k, x_{k+1}, \dots, x_n)}{f_{X_{k+1}, \dots, X_n}(x_{k+1}, \dots, x_n)} 
$$

Para calcular la densidad de probabilidad condicional seguimos los siguientes pasos, los cuáles son análogos al caso discreto:

1. **Obtener la densidad conjunta**: Primero se determina la densidad de probabilidad conjunta de todas las variables involucradas.

2. **Marginalización**: Se calcula la densidad marginal del subconjunto de variables sobre el cual se condiciona. Esto se hace integrando la densidad conjunta sobre las otras variables.

$$
f_Y(y) = \int_{-\infty}^{\infty} f_{X,Y}(x, y) \, dx 
$$

3. **División**: Se divide la densidad conjunta por la densidad marginal para calcular la densidad de $X$ dado que $Y = y$.

$$
f_{X \mid Y}(x \mid y) = \frac{f_{X,Y}(x, y)}{f_Y(y)}
$$


Con lo anterior en mente, procedamos a generar una distribución condicional discreta de una manera más sencilla que encontrar las combinatorias entre los posibles valores de la data. 
1. Seleccionamos un valor $x_i$  desde el cuál queramos hallar la distribución condicional.
2. Generamos el valor $P_X(x_i)$ de la distribución marginal de la variable que condiciona.
3. Tomamos el valor $P_{XY}(x_i,y_j)$ de la matriz de probabilidad conjunta.
4. Efectuamos la división $P_{XY}(x_i,y_j)/P_X(x_i)$
5. Repetimos éste proceso hasta barrer todos los valores $x_i$

Note entonces, que en términos de pandas los procesos anteriores son equivalentes a:
1. Filtrar los datos con respeto al valor $X = x_i$.

````python
# Función para calcular la probabilidad condicional discreta

def probabilidad_condicional_discreta(df, variable_x, variable_y, return_probs=False):
    # Obtener las clases únicas de las variables X e Y
    x_clases = np.sort(df[variable_x].unique())
    y_clases = np.sort(df[variable_y].unique())
    N = len(df)

    # Inicializar matrices para probabilidades conjuntas y condicionales
    P_XY = np.zeros((len(x_clases), len(y_clases)))
    P_Y_dado_X = np.zeros((len(x_clases), len(y_clases)))

    # Calcular la probabilidad conjunta P(X, Y)
    for i, xi in enumerate(x_clases):
        for j, yj in enumerate(y_clases):
            # Calcular P(X = xi, Y = yj)
            P_XY[i, j] = np.sum((df[variable_x] == xi) & (df[variable_y] == yj)) / N
            P_XY[i, j] = np.round(P_XY[i, j], 9)
            
    # Calcular la probabilidad condicional P(Y | X)
    for i in range(len(x_clases)):
        P_X_i = np.sum(P_XY[i, :])
        
        for j in range(len(y_clases)):
            if P_X_i > 0:
                P_Y_dado_X[i, j] = P_XY[i, j] / P_X_i
            else:
                P_Y_dado_X[i, j] = 0
    
    if return_probs:
        P_X = P_XY.sum(axis=1)  # Sumar sobre las filas para obtener P(X)
        P_Y = P_XY.sum(axis=0)  # Sumar sobre las columnas para obtener P(Y)
        return x_clases, y_clases, P_Y_dado_X, P_XY, P_X, P_Y
    else:
        return x_clases, y_clases, P_Y_dado_X
````

````python
# Función para calcular la probabilidad condicional de dos variables contínuas para todo X y Y

def probabilidad_condicional_continua(df, variable_x, variable_y, points=200, bandwidth='scott'):

    # Filtrar las columnas relevantes y eliminar los NaNs
    datos = df[[variable_x, variable_y]].dropna()

    # Estimación de la densidad conjunta utilizando KDE
    kde_conjunta = stats.gaussian_kde(datos.T, bw_method=bandwidth)

    # Estimación de la densidad marginal de X, P(X)
    kde_marginal_X = stats.gaussian_kde(datos[variable_x], bw_method=bandwidth)

    # Creamos una malla de puntos para evaluar la probabilidad condicional
    x = np.linspace(datos[variable_x].min(), datos[variable_x].max(), points)
    y = np.linspace(datos[variable_y].min(), datos[variable_y].max(), points)
    X, Y = np.meshgrid(x, y)
    
    # Para evaluar kde_conjunta, necesitamos pasar las coordenadas como un array de 2 filas
    positions = np.vstack([X.ravel(), Y.ravel()])
    
    # Calculamos la densidad conjunta en cada punto de la malla
    densidad_conjunta = kde_conjunta(positions).reshape(X.shape)
    
    # Calculamos la densidad marginal en cada valor de X
    densidad_marginal_X_values = kde_marginal_X(x)
    densidad_marginal_X = np.tile(densidad_marginal_X_values, (points, 1)).T
    
    # Calculamos la probabilidad condicional P(Y | X)
    Z = densidad_conjunta / densidad_marginal_X

    return X, Y, Z, densidad_conjunta
````

Aplicamos nuestra función de densidad condicionada de probabilidad a nuestras variables de ejemplo y  observamos los gráficos tanto de contorno como 3D.

### **Valor Esperado en una Distribución Multivariante Discreta**

Para un **modelo de probabilidad discreto multivariado**, donde tenemos un vector de variables aleatorias $\mathbf{X} = (X_1, X_2, \ldots, X_n) $, el valor esperado de una función de estas variables se puede calcular para cada variable $X_i$ como la sumatoria ponderada por la probabilidad a lo largo de la variable de interés.

Si queremos calcular el valor esperado de una sola variable $X_i$ en un caso multivariante, la fórmula es:

$$
E[X_i] = \sum_{x_1} \sum_{x_2} \cdots \sum_{x_n} x_i \cdot P(X_1 = x_1, X_2 = x_2, \ldots, X_n = x_n) 
$$

donde $P(X_1 = x_1, X_2 = x_2, \ldots, X_n = x_n) $ es la probabilidad conjunta de que $ X_1 = x_1 $, $ X_2 = x_2 $, ..., $ X_n = x_n $.

Por otra parte, si tenemos una función $ g(X_1, X_2, \ldots, X_n) $ de las variables aleatorias, el valor esperado se calcula como:

$$
E[g(X_1, X_2, \ldots, X_n)] = \sum_{x_1} \sum_{x_2} \cdots \sum_{x_n} g(x_1, x_2, \ldots, x_n) \cdot P(X_1 = x_1, X_2 = x_2, \ldots, X_n = x_n) 

$$

Por ejemplo supongamos que tenemos dos variables aleatorias discretas $X$ y $Y$ con la siguiente tabla de probabilidades conjuntas:

| \( X \ Y \) | 1   | 2   |
|----------------------|-----|-----|
| 1                    | 0.1 | 0.2 |
| 2                    | 0.3 | 0.4 |

Para calcular el valor esperado de $ X $:

$$
E[X] = \sum_{x} \sum_{y} x \cdot P(X = x, Y = y)
$$
$$
E[X] = (1 \cdot (0.1 + 0.2)) + (2 \cdot (0.3 + 0.4))
$$
$$
E[X] = (1 \cdot 0.3) + (2 \cdot 0.7) 
$$
$$
E[X] = 0.3 + 1.4 
$$
$$
E[X] = 1.7 
$$

### **Valor Esperado en una Distribución Multivariante Continua**

Para una **distribución de probabilidad continua multivariante**, donde tenemos un vector de variables aleatorias $ \mathbf{X} = (X_1, X_2, \ldots, X_n) $, para una sola variable $X_i$, el valor esperado se calcula como:

$$
E[X_i] = \int_{-\infty}^{\infty} \cdots \int_{-\infty}^{\infty} x_i \cdot f(x_1, x_2, \ldots, x_n) \, dx_1 \cdots dx_n 
$$

donde $ f(x_1, x_2, \ldots, x_n) $ es la función de densidad conjunta de $ X_1, X_2, \ldots, X_n $.


Para una función $ g(X_1, X_2, \ldots, X_n) $, el valor esperado se calcula como:

$$ 
E[g(X_1, X_2, \ldots, X_n)] = \int_{-\infty}^{\infty} \cdots \int_{-\infty}^{\infty} g(x_1, x_2, \ldots, x_n) \cdot f(x_1, x_2, \ldots, x_n) \, dx_1 \cdots dx_n 
$$

Por ejemplo, supongamos que $X$ y $Y$ siguen una distribución $f_{XY}(x,y)$, el valor esperado de $ X $ es:

$$
E[X] = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} x \cdot f_{XY}(x, y) \, dx \, dy 
$$

### **Correlación de Variables**

La correlación es una medida estadística que indica el grado de *relación lineal entre dos variables*. Específicamente, nos dice qué tan fuerte y en qué dirección (positiva o negativa) se relacionan dos variables. La principal limitación de la correlación radica en que la correlación no implica causalidad. Una fuerte correlación entre dos variables no significa que una cause la otra; puede haber factores subyacentes que influyan en ambas. Además, la correlación de Pearson solo captura relaciones lineales, por lo que no es adecuada para relaciones no lineales. Exísten tres tipos de correlación:

1. **Correlación Positiva**: Ocurre cuando los valores de dos variables tienden a aumentar o disminuir juntos. Un coeficiente de correlación cercano a +1 indica una correlación positiva fuerte.
   
2. **Correlación Negativa**: Ocurre cuando un aumento en una variable tiende a asociarse con una disminución en la otra. Un coeficiente de correlación cercano a -1 indica una correlación negativa fuerte.

3. **Correlación Nula**: Indica que no hay una relación lineal entre las variables. Un coeficiente de correlación cercano a 0 sugiere que no existe una relación lineal.

Antes de entender que es y cómo se calcula la correlación entre variables, necesitamos comprender el concepto de **covarianza**. Aunque no es una métrica de correlación per se, la covarianza es una *medida de cómo dos variables varían conjuntamente*. El cálculo se da mediante: 

$$
\text{Cov}(X,Y) = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{n-1}
$$

Dentro de las métricas más comúnes para calcular la colinealidad entre variables encotramos:

1. **coeficiente de correlación de Pearson**: Es la medida más comúnmente utilizada para evaluar la correlación lineal entre dos variables. Se calcula utilizando la siguiente fórmula, donde el numerador de la correlación corresponde a la covarianza entre variables:

$$
r_{xy} = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}
$$

2. **Correlación de Spearman**: El coeficiente de correlación de Spearman mide la relación monótona entre dos variables. A diferencia de Pearson, que mide relaciones lineales, Spearman evalúa si, al aumentar una variable, la otra también tiende a aumentar o disminuir, pero no necesariamente de manera lineal.Si las variables $X$ y $Y$ se ordenan en rangos $R(X)$ y $R(Y)$, entonces:

$$
\rho_{XY} = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)}
$$

donde:
- $d_i$ es la diferencia entre los rangos $R(X)$ y $R(Y)$ para cada observación $i$.
- $n$ es el número de observaciones.

3. **Correlación de Kendall**: El coeficiente de correlación de Kendall (o tau de Kendall) mide la asociación ordinal entre dos variables. Se basa en la diferencia en el número de pares concordantes y discordantes de observaciones en el conjunto de datos.

$$
\tau = \frac{(C - D)}{\frac{1}{2}n(n-1)}
$$
donde:
- $C$ es el número de pares concordantes.
- $D$ es el número de pares discordantes.


Para ilustrar éstos conceptos, primero generaremos sintéticamente una serie de datos que posean correlación positiva, nula y negativa.

````python
# Ejemplo de correlación sintético
X_pos, Y_pos, X_null, Y_null, X_neg, Y_neg = ejemplo_correlacion()
````

Luego, empleemos el coeficiente de correlación de pearson y la covarianza para evidenciar el valor de la correlación de los tres casos y de su covarianza.

````python
# Función para calcular la covarianza y la correlación entre las variables
def correlacion(X, Y):
    X_mean = np.mean(X)
    Y_mean = np.mean(Y)
    
    covarianza = np.sum((X - X_mean) * (Y - Y_mean)) 
    var_X = np.sum((X - X_mean) ** 2)
    var_Y = np.sum((Y - Y_mean) ** 2)
    
    correlacion = covarianza / np.sqrt(var_X * var_Y)
    return correlacion, covarianza /(len(X)-1)


casos = [[X_pos, Y_pos], [X_null, Y_null], [X_neg, Y_neg]]
for caso in casos:    
    print('*'*36)
    correlacion_val, covarianza_val = correlacion(caso[0], caso[1])
    print(f"Coeficiente de correlación calculado manualmente: {correlacion_val}")
    print(f"Coeficiente de covarianza calculado manualmente: {covarianza_val}")
    print('*'*36)
````

### **Distribución Gaussiana Multivariable**

La **distribución gaussiana multivariable** (o normal multivariada) es una extensión de la distribución normal univariada a múltiples dimensiones. Se caracteriza por un vector de medias y una matriz de covarianza que define la relación entre las variables. La función de densidad de probabilidad para una distribución normal multivariada de $k$ dimensiones es:

$$
f(\mathbf{x}) = \frac{1}{(2\pi)^{k/2} |\Sigma|^{1/2}} \exp\left(-\frac{1}{2} (\mathbf{x} - \mu)^\top \Sigma^{-1} (\mathbf{x} - \mu)\right)
$$

donde:
- $ \mathbf{x} $ es un vector de tamaño $ k $.
- $ \mu $ es el vector de medias.
- $ \Sigma $ es la matriz de covarianza.
- $ |\Sigma| $ es el determinante de $ \Sigma $.

Así pues, los factores más importantes de la distribución gaussiana multivariable son:

1. **Vector de Medias ($\mu$)**:
   En una distribución normal univariada, la media es un valor escalar que representa el centro de la distribución. En una distribución gaussiana multivariable, la media es un **vector** $\mu$ de dimensión $k$, donde $k$ es el número de variables aleatorias. Por ejemplo, si tenemos dos variables $X$ y $Y$, el vector de medias será como se muestra a continuación, donde cada componente representa la media de cada variable aleatoria:
     $$
     \mu = \begin{pmatrix} \mu_1 \\ \mu_2 \end{pmatrix}
     $$

2. **Matriz de Covarianza ($\Sigma$)**:
   La matriz de covarianza $\Sigma$ es una matriz $k \times k$ que captura las **variancias** y **covarianzas** entre las variables. La diagonal de la matriz $\Sigma$ contiene las variancias de cada variable ya que se mide la covarianza entre la variable mísma mientras que los elementos fuera de la diagonal principal capturan la covarianza entre las variables implicadas. Note entonces que la matriz de covarianzas es simétrica y positiva semidefinida:

     $$
     \sigma_{ii} = \text{Var}(X_i)
     $$
     $$
     \sigma_{ij} = \text{Cov}(X_i, X_j)
     $$

La interpretación geométrica consiste en que las varianzas de las variables o diagonal principal de la matriz de covarianzas,  determinan la **extensión** de los elipsoides a lo largo de los ejes principales del espacio (direcciones donde se desarrollan cada variable independientemente) mientras que las covarianzas o elementos no diagonales de la matriz de covarianzas, determinan la **orientación** de los elipsoides, donde las covarianzas distintas de cero inclinan los ejes principales indicando **correlacion** entre las variables.

Para las covarianzas **positivas** la elipsoide se inclina en la dirección en la que ambas variables aumentan juntas. Si las covarianzas son **negativas** la elipsoide se inclina en la dirección en que una variable aumenta mientras la otra disminuye mientras que si las covarianzas son **nulas** el elipsoide se alinea con los ejes principales del espacio indicando independencia entre variables.

### **Distribución Binomial Multivariable**

La **distribución binomial multivariable** (o multinomial) generaliza la distribución binomial a más de dos posibles resultados. Es adecuada para modelar experimentos donde hay más de dos resultados posibles para cada ensayo, como en un lanzamiento de un dado. La función de masa de probabilidad para una distribución multinomial es:

$$
P(X_1 = x_1, X_2 = x_2, \dots, X_k = x_k) = \frac{n!}{x_1! x_2! \dots x_k!} p_1^{x_1} p_2^{x_2} \dots p_k^{x_k}
$$

donde:
- $n $ es el número total de ensayos.
- $x_i $ es el número de veces que el resultado $ i $ ocurre.
- $ p_i $ es la probabilidad del resultado $ i $.

````python
# Definimos el vector de medias y la matriz de covarianza
mu = np.array([0, 0])  # Media en (0, 0)
cov = np.array([[1, 0.5], [0.5, 1]])  # Covarianza
x_ints = [-3,3]
y_ints = [-3,3]

graficar_gauss_3D(mu, cov, x_ints, y_ints)
````

Ahora, para ilustrar la distribución multinomial, generaremos un escenario donde se realizan 10 experimentos donde en cada uno de ellos, exísten tres posibles eventos a ocurrir con diferentes probabilidades de ocurrencia. El codigo siguiente calcula las probabilidades tal que el tercer evento, tenga la restricción $Z = n - X- Y$. Cada punto en el gráfico 3D representa una combinación posible de cuántas veces ocurre cada evento después de los 10 experimentos, y la altura del punto (o la barra en el gráfico) indica la probabilidad de esa combinación específica de resultados según la distribución multinomial. Esta restricción nos permitirá graficar la distribución multinomial en términos de solo las variables $X$ y  $Y$ sin dejar de lado la informazión del evento correspondiente a  $Z$.

$$
P(X = x, Y = y, Z = z) = \frac{n!}{x!y!z!} p_1^x p_2^y p_3^z
$$

Cada punto del gráfico muestra la probabilidad de que, en los $n$ experimentos, ocurran exactamente $X$ veces el primer evento, $Y$ veces el segundo evento, y $Z = n - X - Y$ veces el tercer evento. El gráfico no muestra $Z$ directamente porque no es independiente; está completamente determinado por $X$ y $Y$.

````python
# Parámetros de la distribución multinomial
n = 10  # Número de experimentos
p = [0.3, 0.5, 0.2]  # Probabilidades para cada uno de los 3 eventos

# Crear una malla de puntos para las tres categorías
x = np.arange(0, n+1)
y = np.arange(0, n+1)
X, Y = np.meshgrid(x, y)

# Restricción: z + x + y = n   -> cantidad de experimentos para el tercer evento
Z = n - X - Y

# Filtrar valores válidos (Z >= 0)
valid = (Z >= 0)

# Calculamos la probabilidad multinomial en los puntos válidos
probs = np.zeros_like(X, dtype=float)
for i in range(len(X)):
    for j in range(len(Y)):
        if valid[i, j]:
            z_val = Z[i, j]
            probs[i, j] = (np.math.factorial(n) / 
                           (np.math.factorial(X[i, j]) * np.math.factorial(Y[i, j]) * np.math.factorial(z_val))) * \
                           (p[0] ** X[i, j]) * (p[1] ** Y[i, j]) * (p[2] ** z_val)

# Graficamos la distribución multinomial
graficar_multinomial(X, Y, probs, valid)
````
